```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../dataset/SampleSuperstore.csv", encoding='latin1')
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], format='mixed')
```

```python
df['Month'] = df['Order Date'].dt.to_period('M')
```

```python
monthly_sales = df.groupby('Month')['Sales'].sum()
```

```python
monthly_sales.plot(figsize=(12,5))

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
top_products = df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False).head(10)
```

```python
top_products.plot(kind='bar', figsize=(10,5))

plt.title("Top Selling Products")
plt.xlabel("Sub-Category")
plt.ylabel("Sales")

plt.show()
```

```python
profit_category = df.groupby('Category')['Profit'].sum()
```

```python
profit_category.plot(kind='bar')

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.show()
```

# Insights

1. Technology category generated the highest sales.

2. Some months showed higher revenue compared to others.

3. Furniture category had lower profit than Technology.

4. Sales trends changed over time.
